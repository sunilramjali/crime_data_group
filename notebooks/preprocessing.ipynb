# **1. Import Libraries**


```python
import pandas as pd # to work with dataframes
import numpy as np # to work with arrays and perform mathematical operations
import glob # to import multiple files at once
import os # to create a folder
```

# **2. Data Preprocessing**

**Import csv files**

```python
# Claimant Count
claimant_files = sorted(glob.glob("../data/bronze/*/*claimant_counts*.csv")) # glob.glob returns a list of file paths matching the pattern, sorted to maintain order
claimant = pd.concat([pd.read_csv(file) for file in claimant_files], ignore_index=True) # ignore_index to reset the index after concatenation

# Health
health_files = sorted(glob.glob("../data/bronze/*/*health*.csv"))
health = pd.concat([pd.read_csv(file) for file in health_files], ignore_index=True)
```

**Standardise column names**

```python
dfs = [claimant, health]

for df in dfs:
    df.columns = (
        df.columns
        .str.strip() # remove any leading or trailing whitespace from the column names
        .str.lower() # convert the column names to lowercase
        .str.replace(" ", "_") # replace any spaces in the column names with underscores 
    )
```

```python
claimant.columns
```

```python
health.columns
```

**Drop Unrequired Columns**

```python
claimant = claimant[
    ["area_code", "pop", "claimant_rate"]
]

claimant.columns
```

```python
health = health[
    [col for col in health.columns 
     if "prevalence_rate" in col 
     or col in ["area_code"]]
]

health.columns
```

**Check for Null Values**

```python
display(claimant.isnull().sum(), round(claimant.isnull().mean(), 2)) # check the number of null values in each column and the percentage of null values in each column
```

```python
display(health.isnull().sum(), round(health.isnull().mean(), 2))
```

**Remove duplicates**

```python
# show rows with duplicates
claimant[claimant.duplicated(subset=["area_code"], keep=False)].sort_values(by="area_code").head() # check the rows with duplicate crime_id to understand if there are any patterns in the duplicates
```

```python
# averaging duplicates
claimant = (
    claimant
    .groupby(["area_code", "pop"], as_index=False)
    .agg({"claimant_rate": "mean"})
)

claimant["area_code"].duplicated().sum() # validate 
```

```python
health[health.duplicated(subset=["area_code"], keep=False)].sort_values(by="area_code").head() 
```

```python
health = (
    health
    .groupby("area_code", as_index=False)
    .mean()
)

health["area_code"].duplicated().sum()
```

**Feature Engineering**

```python
# add all the prevelance rate

total_health_prevelance_rate = [col for col in health.columns if col.endswith("_prevalence_rate")]

# Create total column
health["total_prevalence_rate"] = health[total_health_prevelance_rate].sum(axis=1)

# Keep only area_code and total
health = health[
    ["area_code", "total_prevalence_rate"]
]

health.head()
```

**Export**

```python
# Create the directory if it already doesn't exist
os.makedirs("../data/silver", exist_ok=True)

dfs = {
    "claimant_clean": claimant,
    "health_clean": health
}

for name, df in dfs.items():
    df.to_csv(f"../data/silver/{name}.csv", index=False)
```
