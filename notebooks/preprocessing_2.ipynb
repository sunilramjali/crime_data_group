# **2. Data Preprocessing**

**Import csv files**

```python
def load_dataset(pattern):
    """Load csv files from the path you pass into pattern"""
    files = sorted(glob.glob(pattern))

    if not files:
        print(f"No files found for: {pattern}")
        return pd.DataFrame()

    return pd.concat(
        [pd.read_csv(file) for file in files],
        ignore_index=True
    )
```

```python
claimant = load_dataset("../data/bronze/*/*claimant_counts*.csv")
health = load_dataset("../data/bronze/*/*health*.csv")
crime = load_dataset("../data/bronze/*/*crime*.csv")
```

**Standardise column names**

```python
dfs = [claimant, health, crime]

for df in dfs:
    df.columns = (
        df.columns
        .str.strip() # remove any leading or trailing whitespace from the column names
        .str.lower() # convert the column names to lowercase
        .str.replace(" ", "_") # replace any spaces in the column names with underscores 
    )

    print(df.columns)
```

**Drop Unrequired Columns**

```python
claimant_clean = claimant[
    ["area_code", "pop", "claimant_rate"]
]

claimant_clean.columns
```

```python
health_clean = health[
    [col for col in health.columns 
     if "prevalence_rate" in col 
     or col in ["area_code"]]
]

health_clean.columns
```

```python
crime_clean = crime[
    [col for col in crime.columns 
     if "rate" in col 
     or col in ["area_code"]]
]

crime_clean.columns
```

**Check for Null Values**

```python
display(claimant_clean.isnull().sum(), round(claimant_clean.isnull().mean(), 2)) # check the number of null values in each column and the percentage of null values in each column
```

```python
display(health_clean.isnull().sum(), round(health_clean.isnull().mean(), 2))
```

```python
display(crime_clean.isnull().sum(), round(crime_clean.isnull().mean(), 2))
```

**Remove duplicates**

```python
# show rows with duplicates
claimant_clean[claimant_clean.duplicated(subset=["area_code"], keep=False)].sort_values(by="area_code").head() # check the rows with duplicate crime_id to understand if there are any patterns in the duplicates
```

```python
# averaging duplicates
claimant_clean = (
    claimant_clean
    .groupby(["area_code", "pop"], as_index=False)
    .agg({"claimant_rate": "mean"})
)

claimant_clean["area_code"].duplicated().sum() # validate 
```

```python
health_clean[health_clean.duplicated(subset=["area_code"], keep=False)].sort_values(by="area_code").head() 
```

```python
health_clean = (
    health_clean
    .groupby("area_code", as_index=False)
    .mean()
)

health_clean["area_code"].duplicated().sum()
```

```python
crime_clean[crime_clean.duplicated(subset=["area_code"], keep=False)].sort_values(by="area_code").head() 
```

```python
crime_clean = (
    crime_clean
    .groupby("area_code", as_index=False)
    .mean()
)

crime_clean["area_code"].duplicated().sum()
```

**Feature Engineering**

```python
# add all the prevelance rate

total_health_prevelance_rate = [col for col in health_clean.columns if col.endswith("_prevalence_rate")]

# Create total column
health_clean["total_prevalence_rate"] = health_clean[total_health_prevelance_rate].sum(axis=1)

health_clean.head()
```

```python
# add all the crime rate

total_crime_rate = [col for col in crime_clean.columns if col.endswith("_rate")]

# Create total column
crime_clean["total_crime_rate"] = crime_clean[total_crime_rate].sum(axis=1)

crime_clean.head()
```

```python
# Create the directory if it already doesn't exist
os.makedirs("../data/silver", exist_ok=True)

dfs = {
    "claimant_clean": claimant,
    "health_clean": health,
    "crime_clean": crime
}

for name, df in dfs.items():
    df.to_csv(f"../data/silver/{name}.csv", index=False)
```
