```python
#Pull from silver table
mers_agg = spark.table('crime_data.silver.silver_merseyside_crime').toPandas()
```

```python
mers_agg = mers_agg.groupby(['year','month','falls_within','lsoa_code','lsoa_name','crime_type'])['crime_type'].count().reset_index(name='total_crimes')
```

```python
mers_agg = mers_agg.pivot(index = ['year','month','falls_within','lsoa_code','lsoa_name'], columns = 'crime_type', values = 'total_crimes').reset_index()
```

```python
#Since null values in this dataset are not missing, just zero values, they can be replaced with zeros
mers_agg = mers_agg.fillna(0)
```

```python
#Export to gold table
spark.createDataFrame(mers_agg).write.mode('overwrite').saveAsTable('crime_data.gold.gold_merseyside_crime')
```
