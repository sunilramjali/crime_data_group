```python
# ================================================
# INGESTION LAYER
# ================================================
from pyspark.sql.functions import (
    col, count, when, trim, to_date,
    year, month, quarter, date_format,
    row_number, avg, round
)
from pyspark.sql.window import Window

FORCE_NAME = "nottinghamshire"
DATA_PATH = "/Volumes/workspace/default/crime_project/"
OUTPUT_TABLE = f"default.{FORCE_NAME}_reporting_dataset"

df_raw = spark.read \
    .option("header", "true") \
    .option("inferSchema", "false") \
    .option("quote", "\"") \
    .option("escape", "\"") \
    .csv(DATA_PATH) \
    .withColumn("source_file", 
        col("_metadata.file_path"))

print(f"✅ Total rows ingested: {df_raw.count()}")
print(f"✅ Total columns: {len(df_raw.columns)}")
display(df_raw.limit(5))
```

```python
# ================================================
# SAVE RAW INGESTED DATA
# Rename columns first to remove invalid characters
# ================================================

df_raw_clean_cols = df_raw \
    .withColumnRenamed("Crime ID", "crime_id") \
    .withColumnRenamed("Month", "month_year") \
    .withColumnRenamed("Reported by", "force_name") \
    .withColumnRenamed("Falls within", "falls_within") \
    .withColumnRenamed("Longitude", "longitude") \
    .withColumnRenamed("Latitude", "latitude") \
    .withColumnRenamed("Location", "location") \
    .withColumnRenamed("LSOA code", "lsoa_code") \
    .withColumnRenamed("LSOA name", "lsoa_name") \
    .withColumnRenamed("Crime type", "crime_type") \
    .withColumnRenamed("Last outcome category", "outcome") \
    .withColumnRenamed("Context", "context") \
    .withColumn("source_file", col("_metadata.file_path"))

df_raw_clean_cols.write \
    .mode("overwrite") \
    .saveAsTable("default.nottinghamshire_raw")

print("✅ Raw data saved to default.nottinghamshire_raw")
print(f"✅ Total rows saved: {df_raw_clean_cols.count()}")
print(f"✅ Columns: {df_raw_clean_cols.columns}")
```
